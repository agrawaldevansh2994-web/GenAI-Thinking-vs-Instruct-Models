```python
import pandas as pd

final_results = pd.DataFrame({
    "Model": [
        "DeepSeek-R1",
        "GPT-4o mini",
        "GPT-OSS-20B",
        "o3-mini"
    ],
    "GSM8K Accuracy (%)": [89.58, 94.00, 92.00, 92.00],
    "HumanEval Accuracy (%)": [80.00, 92.00, 96.00, 24.00],
    "General Knowledge Accuracy (%)": [100.00, 100.00, 100.00, 100.00]
})

display(final_results.style.hide(axis="index"))
```

## Final Accuracy Calculation

Accuracy was calculated using the formula:

**Accuracy (%) = (Correct Answers / Answered Questions) × 100**

The calculation uses the completed evaluation results from the three tasks. For DeepSeek-R1 on GSM8K, 48 questions were successfully answered because 2 API requests resulted in errors. The remaining finalized task results use their respective completed question counts.

```python
calculation_table = pd.DataFrame({
    "Model": [
        "DeepSeek-R1",
        "GPT-4o mini",
        "GPT-OSS-20B",
        "o3-mini"
    ],
    "GSM8K Calculation": [
        "43 / 48 × 100 = 89.58%",
        "47 / 50 × 100 = 94.00%",
        "46 / 50 × 100 = 92.00%",
        "46 / 50 × 100 = 92.00%"
    ],
    "HumanEval Calculation": [
        "20 / 25 × 100 = 80.00%",
        "23 / 25 × 100 = 92.00%",
        "24 / 25 × 100 = 96.00%",
        "6 / 25 × 100 = 24.00%"
    ],
    "General Knowledge Calculation": [
        "50 / 50 × 100 = 100.00%",
        "50 / 50 × 100 = 100.00%",
        "50 / 50 × 100 = 100.00%",
        "50 / 50 × 100 = 100.00%"
    ]
})

display(calculation_table.style.hide(axis="index"))
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(len(final_results["Model"]))
width = 0.25

plt.figure(figsize=(11, 6))

plt.bar(
    x - width,
    final_results["GSM8K Accuracy (%)"],
    width,
    label="GSM8K"
)

plt.bar(
    x,
    final_results["HumanEval Accuracy (%)"],
    width,
    label="HumanEval"
)

plt.bar(
    x + width,
    final_results["General Knowledge Accuracy (%)"],
    width,
    label="General Knowledge"
)

plt.xticks(x, final_results["Model"])
plt.xlabel("Model")
plt.ylabel("Accuracy (%)")
plt.title("Final Accuracy Comparison Across Tasks")
plt.ylim(0, 110)
plt.legend()
plt.tight_layout()
plt.show()
```

## Final Performance Summary

### Task 1 — GSM8K
GPT-4o mini performed best on GSM8K with an accuracy of 94%.

### Task 2 — HumanEval
GPT-OSS-20B performed best on HumanEval with an accuracy of 96%.

### Task 3 — General Knowledge
All four models achieved 100% accuracy on the General Knowledge task, so there was no single best-performing model based on accuracy.

### Overall Observation
GPT-4o mini showed the highest accuracy on GSM8K, while GPT-OSS-20B performed best on HumanEval. For General Knowledge, all models achieved the same accuracy.
