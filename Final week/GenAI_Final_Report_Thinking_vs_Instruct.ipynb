# Final Report — Thinking Models vs Instruct Models

## 1. Introduction

This project compares Thinking Models and Instruct Models using the same evaluation tasks and common performance metrics.

The evaluation was conducted across three tasks:

- GSM8K — Mathematical Reasoning
- HumanEval — Coding
- General Knowledge — Question & Answer

The four models evaluated were DeepSeek-R1, o3-mini, GPT-4o mini, and GPT-OSS-20B.

The main purpose of the project was to observe differences in accuracy, response time, answer quality, and observable reasoning style across the two model types. The results were collected through API-based testing and analyzed using Python, Excel, and comparison charts.

## 2. Evaluation Methodology

The models were evaluated using the same tasks and comparable evaluation criteria.

### Models Evaluated

- DeepSeek-R1 — Thinking Model
- o3-mini — Thinking Model
- GPT-4o mini — Instruct Model
- GPT-OSS-20B — Instruct Model

GPT-OSS-20B was used as a free replacement for Llama-3.1 8B Instruct because Llama-3.1 8B Instruct was not available in the selected free/developer tier.

### Evaluation Tasks

**Task 1 — GSM8K**

50 mathematical reasoning questions were evaluated. Model answers were compared with the ground-truth answers to calculate accuracy.

**Task 2 — HumanEval**

25 coding problems were evaluated. The generated code was manually checked to determine whether the solution was correct and functional.

**Task 3 — General Knowledge**

50 general knowledge questions were evaluated for answer correctness and clarity.

### Evaluation Metrics

The main metrics used in the project were:

- Accuracy
- Average response time
- Answer quality
- Reasoning quality for Thinking Models

For accuracy, correct responses were compared with the expected answers. Reasoning quality was considered separately from accuracy for the Thinking Models.

### Quality Results

For HumanEval, GPT-OSS-20B achieved the highest Quality Proxy score of 4.92, followed by GPT-4o mini at 4.84, DeepSeek-R1 at 4.28, and o3-mini at 3.48.

The observable responses from DeepSeek-R1 were generally more detailed, while o3-mini responses were generally more concise.

## 5. Key Findings

The results show that model performance depended on the type of task being evaluated.

### Accuracy

- GPT-4o mini achieved the highest GSM8K accuracy at 94%.
- GPT-OSS-20B achieved the highest HumanEval accuracy at 96%.
- All four models achieved 100% accuracy on the General Knowledge task.
- Therefore, no single model achieved the highest accuracy across all three tasks.

### Speed

- GPT-OSS-20B had the lowest average response time in both HumanEval and General Knowledge among the Week 3 tasks.
- DeepSeek-R1 had the highest average response time in both Week 3 tasks.
- This shows a clear difference between the faster Instruct model and the slower Thinking model in these evaluations.

### Quality and Response Style

- GPT-OSS-20B achieved the highest HumanEval Quality Proxy score at 4.92.
- GPT-4o mini also showed a high HumanEval Quality Proxy score of 4.84.
- All four models achieved a Quality Proxy score of 5.0 on General Knowledge.
- DeepSeek-R1 responses were generally more detailed, while o3-mini responses were generally more concise.

Overall, the results suggest that model selection should depend on the task requirements rather than assuming that one model is best for every use case.

## 6. Recommendations

Based on the evaluation results, model selection should depend on the type of task and the required balance between accuracy, speed, and response style.

### GPT-4o mini

GPT-4o mini is a strong choice for a general student chatbot. It achieved the highest GSM8K accuracy at 94%, 92% accuracy on HumanEval, and 100% accuracy on General Knowledge. It also provided a good balance between accuracy, response time, and answer quality.

### GPT-OSS-20B

GPT-OSS-20B is a strong choice when coding performance and fast responses are important. It achieved the highest HumanEval accuracy at 96% and had the lowest average response time in both HumanEval and General Knowledge among the Week 3 tasks.

### DeepSeek-R1

DeepSeek-R1 can be useful when a more detailed reasoning-style response is preferred. Its observable responses were generally more detailed, although its response time was higher than the other models in the Week 3 evaluation.

### o3-mini

o3-mini can be considered when concise responses are preferred. Its observable responses were generally more concise, although its HumanEval accuracy was lower than the other three models in this evaluation.

### Recommendation for a Student Chatbot

For a general-purpose student chatbot, GPT-4o mini would be the recommended choice based on the results of this project. It provided a balanced combination of accuracy, quality, and response time across the evaluated tasks.

However, if the chatbot is mainly intended for coding-related tasks and fast responses, GPT-OSS-20B would be a strong alternative.

## 7. Detailed Task-wise Analysis

### GSM8K — Mathematical Reasoning

The GSM8K evaluation provided a larger mathematical comparison than the initial Week 2 sample. The final results showed that GPT-4o mini achieved the highest accuracy at 94%, followed by GPT-OSS-20B and o3-mini at 92%. DeepSeek-R1 achieved 89.58% based on 48 successfully answered questions.

The results indicate that the Thinking Model category did not automatically produce the highest mathematical accuracy. Although DeepSeek-R1 provided more detailed reasoning-style responses, GPT-4o mini achieved the highest measured accuracy in this task. This shows why model performance should be evaluated using actual benchmark results rather than assumptions based only on model type.

### HumanEval — Coding

HumanEval showed the largest difference in performance among the four models. GPT-OSS-20B achieved the highest accuracy at 96%, followed by GPT-4o mini at 92% and DeepSeek-R1 at 80%. o3-mini achieved 24% accuracy on the selected 25 coding problems.

This task also demonstrated the importance of evaluating generated code separately from general question answering. The coding responses were manually checked to determine whether the solutions were correct and functional. GPT-OSS-20B showed both the highest coding accuracy and the fastest average response time in this evaluation.

### General Knowledge

The General Knowledge task produced the same accuracy result for all four models. Each model achieved 100% accuracy on the selected 50 questions.

Because all models achieved the same accuracy, response time and quality became more useful for comparing their practical performance. GPT-OSS-20B was the fastest model in this task, while all four models received a Quality Proxy score of 5.0.

## 8. Thinking Models vs Instruct Models

The project compared two broad model types: Thinking Models and Instruct Models.

### Thinking Models

The Thinking Models evaluated in this project were DeepSeek-R1 and o3-mini.

These models were observed to provide more reasoning-oriented responses. DeepSeek-R1 generally produced more detailed responses, while o3-mini was generally more concise.

In terms of accuracy, the Thinking Models showed different results across tasks. On GSM8K, DeepSeek-R1 achieved 89.58% and o3-mini achieved 92%. On HumanEval, DeepSeek-R1 achieved 80% while o3-mini achieved 24%. On General Knowledge, both Thinking Models achieved 100%.

### Instruct Models

The Instruct Models evaluated were GPT-4o mini and GPT-OSS-20B.

GPT-4o mini achieved 94% accuracy on GSM8K and 92% on HumanEval. GPT-OSS-20B achieved 92% on GSM8K and the highest HumanEval accuracy of 96%. Both models achieved 100% accuracy on General Knowledge.

The Instruct Models also showed strong response speed in the Week 3 evaluation. GPT-OSS-20B had the lowest average response time for both HumanEval and General Knowledge.

### Overall Comparison

The results show that neither model type was consistently the best across every task.

Thinking Models provided more reasoning-oriented response styles, while Instruct Models showed strong performance in accuracy and response speed in several evaluated tasks.

Therefore, the choice between Thinking and Instruct Models should depend on the specific requirements of the application rather than only on the model category.

## 9. Evaluation Notes and Limitations

The evaluation was conducted using the selected benchmark questions and the available model APIs. The results provide a practical comparison, but some evaluation limitations should be considered when interpreting the findings.

### Different Number of Successfully Answered Questions

For GSM8K, DeepSeek-R1 had 2 API errors during the 50-question evaluation. Therefore, its final accuracy of 89.58% was calculated using 43 correct answers out of 48 successfully answered questions.

The other models completed all 50 GSM8K questions used for their final accuracy calculation.

### Limited HumanEval Sample

The HumanEval evaluation used 25 selected coding problems. The generated code was manually checked for correctness and functionality. Therefore, the results represent the performance of the models on the selected evaluation set rather than every possible coding problem.

### General Knowledge Ceiling Effect

All four models achieved 100% accuracy on the 50 General Knowledge questions. Because every model received the same accuracy score, this task did not provide much differentiation between the models based on accuracy.

Response time and answer quality were therefore more useful for comparing practical performance on this task.

### Quality as an Evaluation Measure

Answer quality was evaluated using a Quality Proxy score. This provides a useful comparison within the project, but quality can be subjective and may vary depending on the evaluator and the criteria used.

### Response Time Variation

Response time represents the average observed time during the evaluation. Actual response time can vary depending on API conditions, network conditions, server load, and other technical factors.

### Model Availability

GPT-OSS-20B was used as a free replacement for Llama-3.1 8B Instruct because Llama-3.1 8B Instruct was not available in the selected free/developer tier.

Therefore, the final comparison is based on the four models that were actually available and evaluated during the project.

### Overall Interpretation

These limitations do not invalidate the results, but they should be considered when drawing conclusions. The purpose of the project was to observe, measure, and compare model performance using the selected tasks and evaluation criteria.

## 10. Conclusion

This project provided a practical comparison between Thinking Models and Instruct Models using mathematical, coding, and general knowledge tasks.

The results showed that model performance varied depending on the task. GPT-4o mini performed best on GSM8K, while GPT-OSS-20B performed best on HumanEval and was also the fastest model in the Week 3 tasks. All four models achieved 100% accuracy on the General Knowledge task.

The evaluation also showed differences in response style. DeepSeek-R1 generally produced more detailed responses, while o3-mini was generally more concise. These differences show that accuracy, speed, quality, and response style should all be considered when selecting a model for a particular application.

Overall, the project demonstrated that there is no single best model for every task. The most suitable model depends on the requirements of the application and the desired balance between accuracy, speed, quality, and response style.
