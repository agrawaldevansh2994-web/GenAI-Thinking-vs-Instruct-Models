- Mathematics

- Planning

### Instruct Models

They are useful for:
- Question answering
- Writing
- Summarization

### Key Differences

Thinking Models:
- More detailed reasoning
- Generally slower
- Suitable for complex problems

### Iterative Approach

```python
def factorial_iterative(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n == 0:
        return 1

    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def factorial_recursive(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

number = 5
print(factorial_iterative(number))
print(factorial_recursive(number))

### RAM vs ROM

**RAM (Random Access Memory)** is the device's **short-term memory**, while **ROM (Read-Only Memory)** stores important instructions that remain available even when the device is powered off.

| Feature | RAM | ROM |
|---|---|---|
| What it does | Stores data and files currently being used | Stores essential instructions needed to start the device |
| Volatility | Volatile — data is lost when power is turned off | Non-volatile — data remains when power is off |
| Speed | Very fast | Generally slower than RAM |
| Capacity | Usually larger, such as 8GB, 16GB or 32GB | Usually much smaller |
| Can you change it? | Yes, data is constantly written and erased | Usually not easily changed; special processes can update it |
| Analogy | Like a whiteboard used while working | Like a permanent cheat sheet |

### In short:

- **RAM** = Working space. More RAM allows you to run more applications smoothly.
- **ROM** = Stores important permanent/startup instructions.

DeepSeek-R1 also noted that the term **ROM** is sometimes used loosely for internal storage in smartphones, although technically true ROM refers to permanent memory used for firmware/startup instructions.

### Initial API Test — GPT-4o mini

The first API test was performed using GPT-4o mini with the question:

"What is 15% of 240?"

The model correctly answered 36.

The recorded response time was approximately 1.9 seconds.

This test confirmed that the OpenAI API connection was working successfully and that model responses and response time could be captured programmatically.

````python
import time

def test_openai_model(question, model="gpt-4o-mini"):
    start_time = time.time()

    response = client.responses.create(
        model=model,
        input=question
    )

    end_time = time.time()

    return {
        "question": question,
        "response": response.output_text,
        "response_time": round(end_time - start_time, 2)
    }
````

````python
result = test_openai_model("What is 15% of 240?")

print("Question:", result["question"])
print("Response:", result["response"])
print("Response Time:", result["response_time"], "seconds")
````

````python
questions = [
    "What is 15% of 240?",
    "What is the capital of France?",
    "Explain photosynthesis in simple words.",
    "Write a Python function to calculate the factorial of a number.",
    "What is the difference between RAM and ROM?",
    "If a train travels 60 km in 2 hours, what is its average speed?",
    "Summarize the importance of teamwork in a workplace.",
    "What are the main advantages of using cloud computing?",
    "Write a Python function to check whether a number is even or odd.",
    "What is the largest planet in our solar system?"
]

print("Total questions:", len(questions))
````

````python
# Thinking vs Instruct Comparison

type_comparison = combined_df.groupby("Model Type").agg(
    Accuracy=("Correct", lambda x: (x == "Correct").mean() * 100),
    Average_Answer_Quality=("Answer Quality (1-5)", "mean"),
    Average_Response_Time=("response_time", "mean")
).reset_index()

type_comparison["Accuracy"] = type_comparison["Accuracy"].round(2)
type_comparison["Average_Answer_Quality"] = type_comparison["Average_Answer_Quality"].round(2)
type_comparison["Average_Response_Time"] = type_comparison["Average_Response_Time"].round(2)

type_comparison
````
